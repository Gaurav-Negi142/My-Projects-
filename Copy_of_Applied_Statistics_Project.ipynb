### Central Tendencies and Dispersion(Explanation)

To get a deeper understanding of our customer data, we need to analyze the central tendencies and dispersion of key variables.

**Measures of Central Tendency** tell us about the typical or central value in a dataset. The main measures are:
- **Mean:** The average of all values.
- **Median:** The middle value when the data is ordered.
- **Mode:** The value that appears most frequently.

**Measures of Dispersion** tell us how spread out the data is. Key measures include:
- **Variance:** The average of the squared differences from the mean.
- **Standard Deviation:** The square root of the variance, providing a measure of the typical distance from the mean.

By calculating and interpreting these measures, we can gain insights into the distribution of customer income, spending habits, and other relevant factors.

```python
# Identify Outliers

# Create boxplots for key variables
key_variables = ['Income', 'Recency', 'MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts',
                 'NumDealsPurchases', 'MntGoldProds', 'NumWebPurchases', 'NumWebVisitsMonth']

for variable in key_variables:
    sns.boxplot(y=df[variable])
    plt.show()
```

```python
# securing percentile so it won't change
original_percentiles = {}
for col in key_variables:
    lower = df[col].quantile(0.05)
    upper = df[col].quantile(0.95)
    original_percentiles[col] = (lower, upper)
```

```python
# Handling outliers
def cap_outliers(df, column, lower_percentile=0.05, upper_percentile=0.95):
    lower = df[column].quantile(lower_percentile)
    upper = df[column].quantile(upper_percentile)
    df[column] = df[column].clip(lower, upper)
    return df

# Apply to all key columns
for col in key_variables:
    df = cap_outliers(df, col)
```

```python
# Rechecking outliers
for col in key_variables:
    lower, upper = original_percentiles[col]
    below = df[df[col] < lower].shape[0]
    above = df[df[col] > upper].shape[0]
    print(f"{col}: Below original 5th = {below}, Above original 95th = {above}")
```

### Identifying and Handling Outliers(Explanation)

Outliers are data points that significantly deviate from other observations. They can heavily influence descriptive statistics like the mean and standard deviation, potentially leading to misleading insights. It's important to identify and handle outliers appropriately to ensure our analysis is robust.

We can identify outliers visually using **box plots**. A box plot shows the distribution of a dataset, including the median, quartiles, and potential outliers (often represented as individual points beyond the "whiskers").

To handle outliers, we use **capping** or **winsorizing**. This involves limiting the extreme values to a certain percentile (e.g., replacing values below the 5th percentile with the 5th percentile value, and values above the 95th percentile with the 95th percentile value). This way, the extreme values are reduced without removing the data points entirely.

In this analysis, we :
1.  **Visualized potential outliers** using box plots for the key numerical variables.
2.  **Identify and cap** the values beyond the 5th and 95th percentiles for these variables to mitigate the impact of outliers.
3.  **Verify** that the capping has been applied by checking the number of values outside the original percentile ranges.

## Task 3 - Probability Distributions 🎲

- **Identify variables that could follow specific probability distributions (e.g., Binomial, Normal). Calculate probabilities and expected values based on these distributions.**



**Deliverables**:

- **Determination of suitable probability distributions for relevant variables and corresponding calculated probabilities and expected values.**:

  

```python

total_purchases = df[['NumDealsPurchases', 'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases']].sum(axis=1)
p_purchase = total_purchases.mean() / total_purchases.max()
expected_purchases = total_purchases.mean()
```

```python
from scipy.stats import norm

mu_income, sigma_income = df['Income'].mean(), df['Income'].std()
prob_income_gt_70000 = 1 - norm.cdf(70000, mu_income, sigma_income)
expected_income = mu_income + sigma_income * norm.ppf(prob_income_gt_70000)

print("Probability Distributions and Calculated Values:")
print(f"Total Purchases:")
print(f"   Probability of a purchase: {p_purchase:.4f}")
print(f"   Expected number of purchases: {expected_purchases:.2f}")

print(f"\nIncome:")
print(f"   Probability of income > $70,000: {prob_income_gt_70000:.4f}")
print(f"   Expected income: {expected_income:.2f}")
```

## Explanation 🎲

After performing the calculations for probability distributions, we've analyzed potential distributions for two key aspects of the customer data:

- **Total Purchases:** By summing up the number of purchases across different channels (deals, web, catalog, store), we calculated the total number of purchases for each customer. We then estimated the probability of a customer making a purchase and the expected number of purchases based on the observed data. This gives us an idea of the typical purchase frequency among our customers.

- **Income:** Assuming the income distribution can be approximated by a Normal distribution (though it's important to note this is a simplification), we calculated the probability of a customer having an income above a specific threshold ($70,000 in this case). We also calculated the expected income based on this probability. This helps us understand the likelihood of finding customers within certain income brackets.

The output displayed above shows the calculated probability of a purchase, the expected number of purchases, the probability of income being greater than $70,000, and the expected income. These values provide initial insights into the distribution of purchasing behavior and income within the dataset.
